```python
def clean_text(text):

  text = str(text).lower()

          # Remove HTML tags
  text = re.sub(r'<.*?>', ' ', text)

                  # Remove punctuation and special characters
  text = re.sub(r'[^a-z\s]', ' ', text)

                          # Tokenization
  words = text.split()

                                  # Remove stopwords and perform lemmatization
  words = [
                                              lemmatizer.lemmatize(word)
                                                      for word in words
                                                              if word not in stop_words
                                                                  ]

  return ' '.join(words)
```
